```python
# Chatbot behavior (instruction)
BOT_PROMPT = """
You are a helpful FAQ chatbot.
Answer user questions using the given FAQ dataset.
Try to match similar questions using simple logic.
If no answer is found, reply:
"Sorry, I don’t have an answer for that yet."
Keep answers short, clear, and professional.
"""
```

```python
# FAQ Data (Questions & Answers)
faq_data = {
    "what is ai": "AI stands for Artificial Intelligence.",
    "what is machine learning": "Machine Learning is a subset of AI.",
    "what is deep learning": "Deep Learning uses neural networks.",
    "what is nlp": "NLP stands for Natural Language Processing.",
    "what is computer vision": "Computer Vision processes images.",
    "what is data science": "Data Science analyzes data.",
    "who are you": "I am a FAQ chatbot.",
    "how are you": "I am working fine!",
    "what can you do": "I answer FAQ questions.",
    "what is python": "Python is a programming language.",
    "what is tensorflow": "TensorFlow is an ML library.",
    "what is pytorch": "PyTorch is a deep learning framework.",
    "what is fastapi": "FastAPI is used to build APIs.",
    "what is github": "GitHub is for code hosting.",
    "what is api": "API connects software systems.",
    "what is chatbot": "A chatbot simulates conversation.",
    "what is llm": "LLM means Large Language Model.",
    "what is cv": "CV means Computer Vision.",
    "what is dl": "DL means Deep Learning.",
    "what is ml": "ML means Machine Learning."
}
```

```python
# Function to compare user input with FAQ questions
def similarity(user_input, question):
    user_words = set(user_input.split())
    question_words = set(question.split())

    # Find matching words
    common = user_words.intersection(question_words)

    # Return score
    return len(common) / len(question_words)
```

```python
# Function to get best answer
def get_response(user_input):

    user_input = user_input.lower()

    best_score = 0
    best_match = None

    # Check all FAQ questions
    for question in faq_data:
        score = similarity(user_input, question)

        if score > best_score:
            best_score = score
            best_match = question

    # If match is good enough
    if best_score > 0.4:
        return faq_data[best_match]

    return "Sorry, I don’t have an answer for that yet."
```

```python
print("🤖 FAQ Chatbot Started (type 'exit' to stop)\n")

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Bot: Goodbye! 👋")
        break

    response = get_response(user_input)
    print("Bot:", response)
```
