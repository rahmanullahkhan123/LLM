```python
from transformers import pipeline
import warnings
warnings.filterwarnings("ignore")

class GrammarCompletionBot:
    def __init__(self, model_name="google/flan-t5-base"):
        print(f"Loading AI Model ({model_name})... Please wait.")
        self.bot = pipeline("text2text-generation", model=model_name)
        print("Model loaded successfully! Type 'exit' or 'quit' to stop.\n")

    def process_input(self, user_text):
        user_text = user_text.strip()
        line_lower = user_text.lower()
        
        if line_lower.startswith("fix:") or line_lower.startswith("correct:"):
            sentence = user_text.split(":", 1)[1].strip()
            prompt = f"Correct the grammar of this sentence: {sentence}"
        elif line_lower.startswith("complete:"):
            sentence = user_text.split(":", 1)[1].strip()
            prompt = f"Complete this sentence logically: {sentence}"
        else:
            prompt = f"Correct the grammar and complete this sentence if it is incomplete: {user_text}"

        response = self.bot(prompt, max_length=150, num_return_sequences=1)
        return response[0]['generated_text']

    def chat(self):
        print("\n🤖 Grammar & Completion AI Chatbot Started!")
        print("-" * 50)
        print("Instructions:")
        print("1. Type 'fix: [your broken sentence]' to correct grammar.")
        print("2. Type 'complete: [your unfinished sentence]' to finish a sentence.")
        print("3. Or just type sentences naturally, and it will try to fix/complete them.")
        print("-" * 50)
        
        while True:
            try:
                user_input = input("You: ")
                if user_input.lower() in ['exit', 'quit', 'stop']:
                    print("Bot: Goodbye! Keep writing beautifully!")
                    break
                
                if not user_input.strip():
                    continue

                response = self.process_input(user_input)
                print(f"Bot: {response}\n")
                
            except KeyboardInterrupt:
                print("\nBot: Goodbye!")
                break

chatbot = GrammarCompletionBot()
chatbot.chat()
```
